```python
def rapikan_status(status):
    # Kalau ada kata "Selesai"
    if 'Selesai' in status:
        return 'Selesai'

    # Kalau ada kata "Batal"
    elif 'Batal' in status:
        return 'Batal'

    # Kalau ada kata "Sedang Dikirim"
    elif 'Sedang Dikirim' in status:
        return 'Dalam Pengiriman'

    # Kalau ada kata "Telah Dikirim"
    elif 'Telah Dikirim' in status:
        return 'Dalam Pengiriman'

    # Kalau ada kata "Pesanan diterima"
    elif 'Pesanan diterima' in status:
        return 'Selesai'

    # Kalau gak ada yang cocok
    else:
        return 'Lainnya'
```

```python
# Test 1
print(rapikan_status("Selesai"))
# Output: Selesai

# Test 2
print(rapikan_status("Batal"))
# Output: Batal

# Test 3
print(rapikan_status("Pesanan diterima, namun Pembeli masih dapat mengajukan pengembalian hingga 2025-12-08"))
# Output: Selesai

# Test 4
print(rapikan_status("Telah Dikirim"))
# Output: Dalam Pengiriman
```

```python
# Cek median dulu
median_qty = df['total_qty'].median()
print("Median total_qty:", median_qty)

# Isi missing dengan median
df['total_qty'] = df['total_qty'].fillna(median_qty)

# Verifikasi
print("Missing setelah isi:", df['total_qty'].isnull().sum())
```

```python
# Isi missing dengan label
df['Metode Pembayaran'] = df['Metode Pembayaran'].fillna('Tidak Diketahui')

# Verifikasi
print("Missing setelah isi:", df['Metode Pembayaran'].isnull().sum())
```

```python
# Cek jumlah baris sebelum
print("Baris sebelum:", len(df))

# Drop baris dengan order_date kosong
df = df.dropna(subset=['order_date'])

# Cek jumlah baris setelah
print("Baris setelah:", len(df))
```

```python
import pandas as pd

df_bersih = pd.read_csv('/content/drive/MyDrive/BELAJAR_DATA/data/indonesia_ecommerce_clean.csv', delimiter=';')

print("Ukuran data bersih:", df_bersih.shape)
print("\nMissing values:")
print(df_bersih.isnull().sum().sum())

print("\nKolom:")
print(df_bersih.columns.tolist())
```

```python
# Bikin ulang kolom Status_Bersih
def rapikan_status(status):
    if 'Selesai' in status:
        return 'Selesai'
    elif 'Batal' in status:
        return 'Batal'
    elif 'Sedang Dikirim' in status:
        return 'Dalam Pengiriman'
    elif 'Telah Dikirim' in status:
        return 'Dalam Pengiriman'
    elif 'Pesanan diterima' in status:
        return 'Selesai'
    else:
        return 'Lainnya'

df_bersih['Status_Bersih'] = df_bersih['Status Pesanan'].apply(rapikan_status)

# Verifikasi
print(df_bersih['Status_Bersih'].value_counts())
print("\nTotal kolom:", len(df_bersih.columns))
```

```python
df_bersih.to_csv('/content/drive/MyDrive/BELAJAR_DATA/data/indonesia_ecommerce_clean.csv', index=False, sep=';')

print("Data bersih disimpan ulang!")
print("Ukuran:", df_bersih.shape)
```

```python
# Hitung proporsi status
status_count = df_bersih['Status_Bersih'].value_counts()
status_pct = df_bersih['Status_Bersih'].value_counts(normalize=True) * 100

# Tampilkan
print("Jumlah per status:")
print(status_count)
print("\nPersentase:")
print(status_pct.round(2))
```

```python
# Lihat alasan pembatalan (hanya untuk pesanan Batal)
batal = df_bersih[df_bersih['Status_Bersih'] == 'Batal']

print("Alasan pembatalan:")
print(batal['Alasan Pembatalan'].value_counts())
```

```python
# Cek pembatalan per metode pembayaran
pivot_bayar = df_bersih.groupby(['Metode Pembayaran', 'Status_Bersih']).size().unstack(fill_value=0)

# Hitung persentase batal
pivot_bayar['% Batal'] = (pivot_bayar.get('Batal', 0) / pivot_bayar.sum(axis=1) * 100).round(2)

print(pivot_bayar)
```

```python
# Top 10 kota dengan pembatalan tertinggi
batal_per_kota = df_bersih[df_bersih['Status_Bersih'] == 'Batal']['Kota/Kabupaten'].value_counts().head(10)

print("Top 10 kota dengan pembatalan terbanyak:")
print(batal_per_kota)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style biar cantik
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# Hitung jumlah per status
status_count = df_bersih['Status_Bersih'].value_counts()

# Bikin grafik
plt.figure(figsize=(8, 5))
sns.barplot(x=status_count.index, y=status_count.values, palette='viridis')

plt.title('Proporsi Status Pesanan', fontsize=14, fontweight='bold')
plt.xlabel('Status')
plt.ylabel('Jumlah Pesanan')

# Tambah label angka di atas bar
for i, v in enumerate(status_count.values):
    plt.text(i, v + 100, f'{v:,}', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Filter hanya pesanan batal
batal = df_bersih[df_bersih['Status_Bersih'] == 'Batal']

# Ambil top 5 alasan
alasan = batal['Alasan Pembatalan'].value_counts().head(5)

# Bikin grafik
plt.figure(figsize=(10, 5))
sns.barplot(x=alasan.values, y=alasan.index, palette='Reds_r')

plt.title('Top 5 Alasan Pembatalan', fontsize=14, fontweight='bold')
plt.xlabel('Jumlah Pembatalan')
plt.ylabel('Alasan')

for i, v in enumerate(alasan.values):
    plt.text(v + 20, i, f'{v:,}', va='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Top 10 kota dengan pembatalan
top_kota = batal['Kota/Kabupaten'].value_counts().head(10)

# Bikin grafik
plt.figure(figsize=(10, 6))
sns.barplot(x=top_kota.values, y=top_kota.index, palette='Oranges_r')

plt.title('Top 10 Kota dengan Pembatalan Terbanyak', fontsize=14, fontweight='bold')
plt.xlabel('Jumlah Pembatalan')
plt.ylabel('Kota')

for i, v in enumerate(top_kota.values):
    plt.text(v + 5, i, f'{v:,}', va='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Pivot: metode pembayaran × status
pivot = df_bersih.groupby(['Metode Pembayaran', 'Status_Bersih']).size().unstack(fill_value=0)

# Urutkan berdasarkan jumlah batal
if 'Batal' in pivot.columns:
    pivot = pivot.sort_values('Batal', ascending=False)

# Ambil top 5
pivot_top = pivot.head(5)

# Bikin grafik
pivot_top.plot(kind='barh', figsize=(10, 6), color=['#2ecc71', '#e74c3c', '#f39c12'])

plt.title('Pembatalan per Metode Pembayaran (Top 5)', fontsize=14, fontweight='bold')
plt.xlabel('Jumlah Pesanan')
plt.ylabel('Metode Pembayaran')
plt.legend(title='Status', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```
